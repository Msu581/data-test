```python
import cv2
import numpy as np
from PIL import Image
from tkinter import Tk, filedialog
import os

# Hide the tkinter window
root = Tk()
root.withdraw()

while True:

    # Select images
    file_paths = filedialog.askopenfilenames(
        title="Select Images",
        filetypes=[
            ("Image Files", "*.jpg *.jpeg *.png *.bmp *.webp"),
            ("All Files", "*.*")
        ]
    )

    if not file_paths:
        print("No images selected.")
        if input("Try again? (Y/N): ").strip().lower() != "y":
            break
        continue

    # Select output folder
    output_folder = filedialog.askdirectory(
        title="Select Folder to Save Images"
    )

    if not output_folder:
        print("No output folder selected.")
        if input("Try again? (Y/N): ").strip().lower() != "y":
            break
        continue

    converted = 0

    for file_path in file_paths:

        try:

            image = cv2.imread(file_path)

            if image is None:
                print(f"Could not read {file_path}")
                continue

            # Convert BGR to RGB
            rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

            # Initial mask
            mask = np.zeros(rgb.shape[:2], np.uint8)

            # Background & Foreground models
            bgdModel = np.zeros((1, 65), np.float64)
            fgdModel = np.zeros((1, 65), np.float64)

            h, w = rgb.shape[:2]

            # Rectangle slightly inside image border
            rect = (10, 10, w - 20, h - 20)

            # Apply GrabCut
            cv2.grabCut(
                rgb,
                mask,
                rect,
                bgdModel,
                fgdModel,
                5,
                cv2.GC_INIT_WITH_RECT
            )

            # Create binary mask
            mask2 = np.where(
                (mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD),
                255,
                0
            ).astype("uint8")

            # Smooth edges
            mask2 = cv2.GaussianBlur(mask2, (5, 5), 0)

            # RGBA image
            rgba = cv2.cvtColor(rgb, cv2.COLOR_RGB2RGBA)
            rgba[:, :, 3] = mask2

            # Crop transparent area
            pil = Image.fromarray(rgba)

            bbox = pil.getbbox()
            if bbox:
                pil = pil.crop(bbox)

            filename = os.path.splitext(os.path.basename(file_path))[0]
            output_path = os.path.join(output_folder, filename + ".png")

            pil.save(output_path)

            converted += 1
            print(f"✓ Saved: {output_path}")

        except Exception as e:
            print(f"✗ Failed: {file_path}")
            print(e)

    print(f"\nSuccessfully processed {converted} image(s).")

    again = input("\nDo you want to process more images? (Y/N): ").strip().lower()

    if again not in ("y", "yes"):
        break

print("\nDone!")
```

```python
from PIL import Image
from tkinter import Tk, filedialog
import os

# Hide the main tkinter window
root = Tk()
root.withdraw()

while True:
    # Select multiple image files
    file_paths = filedialog.askopenfilenames(
        title="Select Images",
        filetypes=[
            ("Image Files", "*.jpg *.jpeg *.png *.bmp *.gif *.tiff *.webp"),
            ("All Files", "*.*")
        ]
    )

    if not file_paths:
        print("No files selected.")

        choice = input("Do you want to try again? (Y/N): ").strip().lower()
        if choice not in ("y", "yes"):
            print("\nThank you for using the Image to ICO Converter!")
            break
        continue

    # Ask where to save the ICO files
    output_folder = filedialog.askdirectory(
        title="Select Folder to Save ICO Files"
    )

    if not output_folder:
        print("No output folder selected.")

        choice = input("Do you want to try again? (Y/N): ").strip().lower()
        if choice not in ("y", "yes"):
            print("\nThank you for using the Image to ICO Converter!")
            break
        continue

    converted = 0

    for file_path in file_paths:
        try:
            img = Image.open(file_path).convert("RGBA")

            # Ask for the name of this image
            default_name = os.path.splitext(os.path.basename(file_path))[0]

            new_name = input(
                f"\nEnter name for '{default_name}' (leave blank to keep original): "
            ).strip()

            if not new_name:
                new_name = default_name

            output_file = os.path.join(output_folder, new_name + ".ico")

            # Prevent overwriting existing files
            counter = 1
            while os.path.exists(output_file):
                output_file = os.path.join(
                    output_folder,
                    f"{new_name}_{counter}.ico"
                )
                counter += 1

            # Standard Windows icon sizes
            sizes = [
                (16, 16),
                (24, 24),
                (32, 32),
                (48, 48),
                (64, 64),
                (128, 128),
                (256, 256)
            ]

            img.save(output_file, format="ICO", sizes=sizes)
 
            converted += 1
            print(f"✓ Saved as: {os.path.basename(output_file)}")

        except Exception as e:
            print(f"✗ Failed: {file_path}")
            print(e)

    print(f"\nSuccessfully converted {converted} image(s).")

    # Ask if the user wants to convert more images
    while True:
        choice = input("\nDo you want to convert more images? (Y/N): ").strip().lower()

        if choice in ("y", "yes"):
            print()
            break   # Continue outer while loop

        elif choice in ("n", "no"):
            print("\nThank you for using the Image to ICO Converter!")
            break

        else:
            print("Please enter Y or N.")
```
